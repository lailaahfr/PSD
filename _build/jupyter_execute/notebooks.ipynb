# Polutan Kabupaten Bangkalan

## Latar Belakang

Kualitas udara merupakan salah satu aspek lingkungan yang penting untuk diperhatikan karena berhubungan dengan kondisi lingkungan dan aktivitas masyarakat. Udara dapat mengandung berbagai jenis polutan yang berasal dari aktivitas manusia maupun proses alami. Beberapa polutan yang umum digunakan untuk menggambarkan kondisi kualitas udara antara lain karbon monoksida (CO), nitrogen dioksida (NO₂), sulfur dioksida (SO₂), ozon (O₃), serta partikulat seperti PM2.5 dan PM10.

Kabupaten Bangkalan dipilih sebagai wilayah kajian karena merupakan salah satu wilayah di Pulau Madura yang memiliki aktivitas masyarakat, transportasi, perdagangan, dan kegiatan ekonomi lainnya. Berbagai aktivitas tersebut berpotensi menghasilkan emisi yang dapat memengaruhi kondisi kualitas udara. Namun, kondisi kualitas udara suatu wilayah tidak dapat ditentukan hanya berdasarkan asumsi, sehingga diperlukan data untuk melihat perubahan konsentrasi polutan secara lebih objektif.

Pada penelitian ini, data kualitas udara di Kabupaten Bangkalan dikumpulkan berdasarkan lokasi dan waktu pengamatan. Wilayah pengambilan data dibatasi menggunakan Area of Interest (AOI) berdasarkan koordinat wilayah Kabupaten Bangkalan. Data yang dikumpulkan mencakup periode minimal satu tahun, yaitu hingga 30 Agustus 2026.

Data tersebut selanjutnya dipahami dan dieksplorasi untuk mengetahui karakteristik, kelengkapan, serta kualitas data sebelum digunakan dalam analisis lebih lanjut. Hasil pengolahan data akan digunakan untuk melihat perubahan konsentrasi polutan dari waktu ke waktu melalui visualisasi time series.
