# Polutan di Kabupaten Bangkalan, Jawa Timur  
*(Data Sentinel‑5P via Copernicus Data Space & openEO)*

Proyek ini menganalisis polutan udara di Kabupaten Bangkalan, Jawa Timur, menggunakan data pengamatan satelit **Sentinel‑5P** yang diakses melalui **Copernicus Data Space Ecosystem** dan layanan **openEO**. Fokus utama adalah polutan kunci seperti **nitrogen dioksida (NO₂)**, **sulfur dioksida (SO₂)**, dan **karbon monoksida (CO)** yang umum dipakai untuk menggambarkan kondisi polutan udara.
